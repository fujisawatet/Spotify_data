# 説明書 Markdown → Word 変換ノートブック

**手順**:
1. セル1を実行（Pandocをインストール）
2. セル2を実行（ファイルをアップロード）→ `Phase1_説明書.md` と `Phase2_説明書.md` を選択
3. セル3を実行（変換・ダウンロード）

```python
# セル1: Pandocをインストール（1〜2分）
!apt-get install -y pandoc > /dev/null 2>&1
!pandoc --version | head -1
print("Pandoc インストール完了")
```

```python
# セル2: ファイルをアップロード
# Phase1_説明書.md と Phase2_説明書.md の2ファイルを選択してください
from google.colab import files
uploaded = files.upload()
print(f"アップロードされたファイル: {list(uploaded.keys())}")
```

```python
# セル3: Word形式に変換してダウンロード
import subprocess
from google.colab import files

targets = [
    ("Phase1_説明書.md", "Phase1_説明書.docx"),
    ("Phase2_説明書.md", "Phase2_説明書.docx"),
]

for md_file, docx_file in targets:
    if md_file in uploaded:
        result = subprocess.run(
            ["pandoc", md_file, "-o", docx_file,
             "--highlight-style=tango"],
            capture_output=True, text=True
        )
        if result.returncode == 0:
            print(f"✓ {docx_file} を作成しました → ダウンロード開始")
            files.download(docx_file)
        else:
            print(f"✗ {md_file} の変換に失敗しました")
            print(result.stderr)
    else:
        print(f"⚠ {md_file} がアップロードされていません")
```
